```python
import warnings
from collections import Counter
from pathlib import Path

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score
from sklearn.preprocessing import LabelEncoder

warnings.filterwarnings('ignore')
sns.set_style('whitegrid')

ROOT_DIR = Path(r'D:\Projects\Research_Component')
DATA_DIR = ROOT_DIR / 'data'
PROC_DIR = DATA_DIR / 'processed'
MODEL_DIR = ROOT_DIR / 'models'
PLOT_DIR = ROOT_DIR / 'plots' / '09_bootstrap'
PLOT_DIR.mkdir(parents=True, exist_ok=True)

STAGE2_MODEL_PATH = MODEL_DIR / 'stage2_rf_model.pkl'
STAGE2_ENCODER_PATH = MODEL_DIR / 'stage2_label_encoder.pkl'
STAGE2_FEATURES_PATH = MODEL_DIR / 'stage2_feature_cols.pkl'
SEVERITY_FILE = PROC_DIR / 'daph_severity_dataset.xlsx'
FEATURE_FILE = PROC_DIR / 'FMD_model_ready_main refined_final_dataset.csv'

stage2_rf_model = joblib.load(STAGE2_MODEL_PATH)
stage2_label_encoder = joblib.load(STAGE2_ENCODER_PATH)
stage2_feature_cols = list(joblib.load(STAGE2_FEATURES_PATH))

severity_raw = pd.read_excel(SEVERITY_FILE, sheet_name='DAPH Data', header=1)
feature_df = pd.read_csv(FEATURE_FILE)

print('Setup complete')
print(f'Stage 2 model type      : {type(stage2_rf_model).__name__}')
print(f'Severity workbook shape : {severity_raw.shape}')
print(f'Feature dataset shape   : {feature_df.shape}')
print(f'Stage 2 feature count    : {len(stage2_feature_cols)}')
if hasattr(stage2_label_encoder, 'classes_'):
    print(f'Label encoder classes    : {list(stage2_label_encoder.classes_)}')
else:
    print(f'Label encoder mapping    : {stage2_label_encoder}')
```

```python
severity_df = severity_raw.copy()
severity_df.columns = severity_df.columns.str.strip()

rename_map = {
    'District': 'district',
    'Year': 'year',
    'Cases': 'cases',
    'Deaths': 'deaths',
    'Outbreak_Months': 'outbreak_months',
    'Outbreak Months': 'outbreak_months',
    'Months': 'outbreak_months',
}
severity_df = severity_df.rename(columns=rename_map)

required_cols = ['district', 'year', 'cases', 'deaths', 'outbreak_months']
missing_cols = [col for col in required_cols if col not in severity_df.columns]
if missing_cols:
    raise ValueError(f'Missing required DAPH columns: {missing_cols}')

severity_df['district'] = severity_df['district'].astype(str).str.strip()
severity_df['year'] = pd.to_numeric(severity_df['year'], errors='coerce')
severity_df['cases'] = pd.to_numeric(severity_df['cases'], errors='coerce')
severity_df['deaths'] = pd.to_numeric(severity_df['deaths'], errors='coerce')
severity_df['outbreak_months'] = pd.to_numeric(severity_df['outbreak_months'], errors='coerce')
severity_df = severity_df.dropna(subset=required_cols).copy()
severity_df['year'] = severity_df['year'].astype(int)

severity_df['severity_score'] = severity_df['cases'] * (severity_df['outbreak_months'] / 12.0) + severity_df['deaths'] * 10

def score_to_class(score):
    if score < 50:
        return 'LOW'
    if score < 300:
        return 'MEDIUM'
    return 'HIGH'

severity_df['severity_class'] = severity_df['severity_score'].apply(score_to_class)
severity_lookup = severity_df[['district', 'year', 'severity_score', 'severity_class']].drop_duplicates(subset=['district', 'year']).copy()

feature_df_clean = feature_df.copy()
feature_df_clean['district'] = feature_df_clean['district'].astype(str).str.strip()
feature_df_clean['year'] = pd.to_numeric(feature_df_clean['year'], errors='coerce')
feature_df_clean = feature_df_clean.dropna(subset=['district', 'year']).copy()
feature_df_clean['year'] = feature_df_clean['year'].astype(int)

merged_df = feature_df_clean.merge(severity_lookup[['district', 'year', 'severity_class']], on=['district', 'year'], how='left')
merged_df = merged_df[merged_df['Outbreak status'] == 1].copy()
merged_df = merged_df.dropna(subset=['severity_class']).reset_index(drop=True)

print(f'Total outbreak rows with severity labels: {len(merged_df)}')
print('Severity class distribution:')
print(merged_df['severity_class'].value_counts().reindex(['LOW', 'MEDIUM', 'HIGH'], fill_value=0))
print('Years available:', sorted(merged_df['year'].unique().tolist()))
```

```python
severity_to_code = {'LOW': 0, 'MEDIUM': 1, 'HIGH': 2}
code_to_severity = {value: key for key, value in severity_to_code.items()}
merged_df['severity_code'] = merged_df['severity_class'].map(severity_to_code).astype(int)

def decode_severity_code(code):
    return code_to_severity[int(code)]

def bootstrap_predict(X_train, y_train, X_test, n_iterations=100, random_state=42):
    X_train = np.asarray(X_train)
    y_train = np.asarray(y_train).astype(int)
    X_test = np.asarray(X_test)

    rng = np.random.default_rng(random_state)
    predictions_list = []

    for i in range(n_iterations):
        if (i + 1) % 10 == 0 or i == 0:
            print(f'Iteration {i + 1}/{n_iterations}')

        indices = rng.choice(len(X_train), size=len(X_train), replace=True)
        X_boot = X_train[indices]
        y_boot = y_train[indices]

        if np.unique(y_boot).size < 2:
            continue

        model = RandomForestClassifier(
            n_estimators=100,
            class_weight='balanced',
            random_state=i,
            n_jobs=-1
        )
        model.fit(X_boot, y_boot)
        predictions_list.append(model.predict(X_test).astype(int))

    if not predictions_list:
        raise ValueError('Bootstrap produced no valid models.')

    prediction_matrix = np.vstack(predictions_list)
    rows = []

    for sample_idx in range(prediction_matrix.shape[1]):
        votes = prediction_matrix[:, sample_idx]
        counts = Counter(votes)
        most_common_prediction, most_common_count = counts.most_common(1)[0]

        if len(set(votes)) == 1:
            low_code = high_code = int(votes[0])
        else:
            low_code = int(np.clip(np.round(np.percentile(votes, 5)), 0, 2))
            high_code = int(np.clip(np.round(np.percentile(votes, 95)), 0, 2))

        rows.append({
            'prediction': decode_severity_code(most_common_prediction),
            'low_bound': decode_severity_code(low_code),
            'high_bound': decode_severity_code(high_code),
            'confidence_pct': round((most_common_count / len(votes)) * 100, 1),
        })

    return pd.DataFrame(rows)

print('Bootstrap function defined successfully')
```

```python
valid_test_years = [2018, 2019, 2021, 2022]
loyo_results = []
loyo_detailed = {}

for test_year in valid_test_years:
    train_mask = merged_df['year'] != test_year
    test_mask = merged_df['year'] == test_year

    X_train = merged_df.loc[train_mask, stage2_feature_cols].fillna(0).astype(float).to_numpy()
    y_train = merged_df.loc[train_mask, 'severity_code'].to_numpy()
    X_test = merged_df.loc[test_mask, stage2_feature_cols].fillna(0).astype(float).to_numpy()
    y_test = merged_df.loc[test_mask, 'severity_code'].to_numpy()

    print(f'\nYear {test_year} | Test samples: {len(X_test)}')
    bootstrap_df = bootstrap_predict(X_train, y_train, X_test, n_iterations=100, random_state=42)
    bootstrap_df['true_code'] = y_test
    bootstrap_df['true_severity'] = [decode_severity_code(code) for code in y_test]
    bootstrap_df['year'] = test_year
    bootstrap_df['prediction_code'] = bootstrap_df['prediction'].map(severity_to_code)
    bootstrap_df['low_code'] = bootstrap_df['low_bound'].map(severity_to_code)
    bootstrap_df['high_code'] = bootstrap_df['high_bound'].map(severity_to_code)
    bootstrap_df['interval_width'] = bootstrap_df['high_code'] - bootstrap_df['low_code']

    loyo_detailed[test_year] = bootstrap_df

    print(bootstrap_df[['prediction', 'low_bound', 'high_bound', 'confidence_pct']].head(5).to_string(index=False))
```

```python
evaluation_rows = []

for test_year in valid_test_years:
    fold_df = loyo_detailed[test_year]
    accuracy_pct = (fold_df['prediction_code'] == fold_df['true_code']).mean() * 100
    coverage_pct = ((fold_df['true_code'] >= fold_df['low_code']) & (fold_df['true_code'] <= fold_df['high_code'])).mean() * 100
    mean_confidence_pct = fold_df['confidence_pct'].mean()

    evaluation_rows.append({
        'Year': test_year,
        'Samples': len(fold_df),
        'Accuracy': accuracy_pct,
        'Coverage': coverage_pct,
        'Mean Confidence': mean_confidence_pct,
    })

eval_df = pd.DataFrame(evaluation_rows)
mean_row = {
    'Year': 'Mean',
    'Samples': '-',
    'Accuracy': eval_df['Accuracy'].mean(),
    'Coverage': eval_df['Coverage'].mean(),
    'Mean Confidence': eval_df['Mean Confidence'].mean(),
}
summary_df = pd.concat([eval_df, pd.DataFrame([mean_row])], ignore_index=True)

display_df = summary_df.copy()
for col in ['Accuracy', 'Coverage', 'Mean Confidence']:
    display_df[col] = display_df[col].apply(lambda value: f'{value:.1f}%')

print('Bootstrap LOYO Summary')
print('=' * 78)
print(display_df[['Year', 'Samples', 'Accuracy', 'Coverage', 'Mean Confidence']].to_string(index=False))
print('\nCoverage should ideally be at least 80% for an acceptable interval.')
```

```python
bootstrap_all_confidences = pd.concat([fold_df['confidence_pct'] for fold_df in loyo_detailed.values()], ignore_index=True)
bootstrap_all_intervals = pd.concat([fold_df[['true_severity', 'interval_width']] for fold_df in loyo_detailed.values()], ignore_index=True)

fig, ax = plt.subplots(figsize=(10, 6))
sns.histplot(bootstrap_all_confidences, bins=20, kde=True, color='#1f77b4', ax=ax)
ax.set_title('Bootstrap Prediction Confidence Distribution')
ax.set_xlabel('Confidence (%)')
ax.set_ylabel('Count of Predictions')
fig.tight_layout()
confidence_plot_path = PLOT_DIR / 'confidence_distribution.png'
fig.savefig(confidence_plot_path, dpi=150, bbox_inches='tight')
plt.close(fig)

fig, ax = plt.subplots(figsize=(9, 6))
sns.boxplot(data=bootstrap_all_intervals, x='true_severity', y='interval_width', order=['LOW', 'MEDIUM', 'HIGH'], palette='Set2', ax=ax)
ax.set_title('Prediction Interval Width by Severity Class')
ax.set_xlabel('True Severity Class')
ax.set_ylabel('Interval Width (ordinal steps)')
fig.tight_layout()
width_plot_path = PLOT_DIR / 'interval_width_by_class.png'
fig.savefig(width_plot_path, dpi=150, bbox_inches='tight')
plt.close(fig)

coverage_plot_df = eval_df.copy()
fig, ax = plt.subplots(figsize=(9, 6))
ax.bar(coverage_plot_df['Year'].astype(str), coverage_plot_df['Coverage'], color='#2ca02c')
ax.axhline(80, color='red', linestyle='--', linewidth=2, label='80% threshold')
ax.set_title('Bootstrap Interval Coverage by Year')
ax.set_xlabel('Test Year')
ax.set_ylabel('Coverage (%)')
ax.legend()
fig.tight_layout()
coverage_plot_path = PLOT_DIR / 'coverage_by_year.png'
fig.savefig(coverage_plot_path, dpi=150, bbox_inches='tight')
plt.close(fig)

print(f'Saved: {confidence_plot_path}')
print(f'Saved: {width_plot_path}')
print(f'Saved: {coverage_plot_path}')
```

```python
def bootstrap_predict_with_votes(X_train, y_train, X_test, n_iterations=100, random_state=42):
    X_train = np.asarray(X_train)
    y_train = np.asarray(y_train).astype(int)
    X_test = np.asarray(X_test)

    rng = np.random.default_rng(random_state)
    predictions_list = []

    for i in range(n_iterations):
        if (i + 1) % 10 == 0 or i == 0:
            print(f'Iteration {i + 1}/{n_iterations}')

        indices = rng.choice(len(X_train), size=len(X_train), replace=True)
        X_boot = X_train[indices]
        y_boot = y_train[indices]

        if np.unique(y_boot).size < 2:
            continue

        model = RandomForestClassifier(
            n_estimators=100,
            class_weight='balanced',
            random_state=i,
            n_jobs=-1,
        )
        model.fit(X_boot, y_boot)
        predictions_list.append(model.predict(X_test).astype(int))

    if not predictions_list:
        raise ValueError('Bootstrap produced no valid models.')

    prediction_matrix = np.vstack(predictions_list)
    rows = []

    for sample_idx in range(prediction_matrix.shape[1]):
        votes = prediction_matrix[:, sample_idx]
        counts = Counter(votes)
        most_common_prediction, most_common_count = counts.most_common(1)[0]

        if len(set(votes)) == 1:
            low_code = high_code = int(votes[0])
        else:
            low_code = int(np.clip(np.round(np.percentile(votes, 5)), 0, 2))
            high_code = int(np.clip(np.round(np.percentile(votes, 95)), 0, 2))

        rows.append(
            {
                'prediction': decode_severity_code(most_common_prediction),
                'low_bound': decode_severity_code(low_code),
                'high_bound': decode_severity_code(high_code),
                'confidence_pct': round((most_common_count / len(votes)) * 100, 1),
            }
        )

    return pd.DataFrame(rows), prediction_matrix


final_train_mask = merged_df['year'] <= 2023
X_train_final = merged_df.loc[final_train_mask, stage2_feature_cols].fillna(0).astype(float).to_numpy()
y_train_final = merged_df.loc[final_train_mask, 'severity_code'].to_numpy()

outbreak_all = feature_df_clean[feature_df_clean['Outbreak status'] == 1].copy()
outbreak_all = outbreak_all.merge(
    severity_lookup[['district', 'year', 'severity_class']],
    on=['district', 'year'],
    how='left',
)
outbreak_all = outbreak_all.reset_index(drop=True)

X_test_all = outbreak_all[stage2_feature_cols].fillna(0).astype(float).to_numpy()
bootstrap_summary_all, bootstrap_votes_all = bootstrap_predict_with_votes(
    X_train_final,
    y_train_final,
    X_test_all,
    n_iterations=100,
    random_state=42,
)

bootstrap_results_df = pd.DataFrame(
    {
        'district': outbreak_all['district'],
        'year': outbreak_all['year'],
        'month_num': outbreak_all['month_num'],
        'true_severity': outbreak_all['severity_class'],
        'predicted_severity': bootstrap_summary_all['prediction'],
        'interval_low': bootstrap_summary_all['low_bound'],
        'interval_high': bootstrap_summary_all['high_bound'],
        'confidence_pct': bootstrap_summary_all['confidence_pct'],
    }
)
bootstrap_results_df['interval_label'] = bootstrap_results_df.apply(
    lambda row: f"[{row['interval_low']}, {row['interval_high']}]",
    axis=1,
)

bootstrap_csv_path = PROC_DIR / 'bootstrap_intervals.csv'
bootstrap_results_df.to_csv(bootstrap_csv_path, index=False)

print(f'Bootstrap results shape: {bootstrap_results_df.shape}')
print('\nSample rows:')
print(bootstrap_results_df.head(10).to_string(index=False))

interval_low_codes = bootstrap_results_df['interval_low'].map(severity_to_code).astype(int)
interval_high_codes = bootstrap_results_df['interval_high'].map(severity_to_code).astype(int)
interval_width_codes = interval_high_codes - interval_low_codes
narrow_count = int((interval_width_codes == 0).sum())
medium_count = int((interval_width_codes == 1).sum())
wide_count = int((interval_width_codes == 2).sum())

print('\nDistribution of interval widths:')
print(f'Narrow  [X, X]     = {narrow_count}')
print(f'Medium  [LOW, MED] = {medium_count}')
print(f'Wide    [LOW, HIGH]= {wide_count}')
print(f'\nSaved bootstrap intervals to: {bootstrap_csv_path}')
```

```python
case_row = outbreak_all[(outbreak_all['district'] == 'Anuradhapura') & (outbreak_all['month_num'] == 1) & (outbreak_all['year'] == 2024)]

if case_row.empty:
    raise ValueError('Anuradhapura January 2024 row was not found in the outbreak dataset.')

X_case = case_row[stage2_feature_cols].fillna(0).astype(float).to_numpy()
case_summary, case_votes = bootstrap_predict_with_votes(
    X_train_final,
    y_train_final,
    X_case,
    n_iterations=100,
    random_state=42,
)
case_summary_row = case_summary.iloc[0]
case_votes_flat = case_votes[:, 0]
case_vote_counts = Counter(case_votes_flat)

low_votes = int(case_vote_counts.get(severity_to_code['LOW'], 0))
medium_votes = int(case_vote_counts.get(severity_to_code['MEDIUM'], 0))
high_votes = int(case_vote_counts.get(severity_to_code['HIGH'], 0))

confidence_pct = float(case_summary_row['confidence_pct'])
if confidence_pct >= 70:
    confidence_label = 'High'
elif confidence_pct >= 50:
    confidence_label = 'Moderate'
else:
    confidence_label = 'Low'

print('BOOTSTRAP UNCERTAINTY REPORT')
print('=' * 29)
print('District     : Anuradhapura')
print('Month        : January 2024')
print('')
print('Point Estimate:')
print(f'  Predicted severity  : {case_summary_row["prediction"]}')
print('')
print('Bootstrap Results (100 iterations):')
print(f'  LOW predictions     : {low_votes}/100')
print(f'  MEDIUM predictions  : {medium_votes}/100')
print(f'  HIGH predictions    : {high_votes}/100')
print('')
print('95% Prediction Interval:')
print(f'  Lower bound         : {case_summary_row["low_bound"]}')
print(f'  Upper bound         : {case_summary_row["high_bound"]}')
print(f'  Interval            : [{case_summary_row["low_bound"]}, {case_summary_row["high_bound"]}]')
print('')
print(f'Confidence: {confidence_pct:.0f}% of iterations agreed on {case_summary_row["prediction"]}')
print('')
print('Interpretation:')
print(f'  Model predicts {case_summary_row["prediction"]} severity with {confidence_label.lower()} confidence. Outcome could range from {case_summary_row["low_bound"]} to {case_summary_row["high_bound"]}.')
```
